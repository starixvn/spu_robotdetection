```python
for k in range(2,31):
    check_files_sameday = [f for f in sorted(os.listdir(root_orig))]
    myStrCheck = check_files_sameday[0].rsplit("_",4)
    pattern = "^" + myStrCheck[0]
    results = [i for i in check_files_sameday if re.findall(pattern,i)]
    
    data_allcus = pd.concat((pd.read_csv(os.path.join(root_orig, f),encoding='ISO-8859-1', skiprows=[1], skipfooter=1).assign(file_name = f) for f in results))
    
    data_allcus.to_csv(f'D:\Project\DetectRobot\Data\data_month6_eachday\data_{k}06.csv')
    
    for file in results:
        os.rename(os.path.join(from_location,file),os.path.join(to_location,file))
```
